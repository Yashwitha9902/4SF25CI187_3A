Selecting a dataset and reading the CSV file using Pandas.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("spotify_tracks.csv")
```

the dataset information

```python

print(df.info())
```

STATISTICAL SUMMARY :

```python

print(df.describe())
```

Removing unnecessary columns(DROP)

```python

df = df.drop(columns=["Title"])
```

 Converting Artist and Track columns into string data type

```python

df["Artist"] = df["Artist"].astype("string")
df["Track"] = df["Track"].astype("string")
```

Adding a new column for song duration in seconds

```python
df["Duration_sec"] = df["Duration_min"] * 60
```

UPDATED DATASET


```python

print(df.head())
```

 updated data types

```python

print(df.dtypes)
```

Checking the MISSING VALUES

```python

print(df.isnull().sum())
```

 Selecting the numerical columns

```python

numeric_columns = df.select_dtypes(include="number").columns

```


 Replacing the missing values with the mean of each column

```python

for col in numeric_columns:
    df[col] = df[col].fillna(df[col].mean())
```

Checking the missing values again

```python

print("Missing values after imputation:")
print(df.isnull().sum())
```

Arranging the songs based on Views in descending order (SORTING)

```python

top_songs = df.sort_values(by="Views", ascending=False).head(10)
```

 the top 10 songs

```python

print(top_songs[["Track", "Artist", "Views"]])
```

Arranging the songs based on Stream in descending order (SORTING)

```python

top_streamed = df.sort_values(by="Stream", ascending=False).head(10)
```

 the top 10 streamed songs

```python

print(top_streamed[["Track", "Artist", "Stream"]])
```

AVERAGE DANCEABILITY

```python

avg_danceability = df["Danceability"].mean()
```

AVERAGE ENERGY

```python

avg_energy = df["Energy"].mean()
```

```python
print("Average Danceability:", avg_danceability)
print("Average Energy:", avg_energy)
```

Grouping the songs by artist and calculating the TOTAL VIEWS

```python

artist_views = df.groupby("Artist")["Views"].sum()
```

Arranging the artists by total views (SORTING)

```python

top_artists = artist_views.sort_values(ascending=False).head(10)
```

TOP ARTISTS

```python

print(top_artists)
```

AVERAGE LIKES

```python

average_likes = df["Likes"].mean()

print("Average Likes:", average_likes)
```

top 10 songs based on views

```python

top_songs = df.sort_values(
    by="Views", ascending=False
).head(10)
```

top 10 most-viewed songs using a BAR CHART

```python
# Create bar chart
plt.figure(figsize=(10, 5))
plt.bar(top_songs["Track"], top_songs["Views"])

plt.xlabel("Song Name")
plt.ylabel("Number of Views")
plt.title("Top 10 Most Viewed Songs")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

distribution of song energy using a HISTOGRAM

```python
plt.figure(figsize=(8, 5))

plt.hist(df["Energy"], bins=20)

plt.xlabel("Energy")
plt.ylabel("Number of Songs")
plt.title("Distribution of Song Energy")

plt.show()
```

relationship between danceability and energy using a SCATTER PLOT

```python
plt.figure(figsize=(8, 5))
plt.scatter(df["Danceability"], df["Energy"])

plt.xlabel("Danceability")
plt.ylabel("Energy")
plt.title("Danceability vs Energy")

plt.show()
```

top 10 artists based on total views using a BAR CHART

```python
# Calculate total views for each artist
artist_views = df.groupby("Artist")["Views"].sum()

# Select top 10 artists
top_artists = artist_views.sort_values(
    ascending=False
).head(10)

# Create bar chart
plt.figure(figsize=(10, 5))

plt.bar(top_artists.index, top_artists.values)

plt.xlabel("Artist")
plt.ylabel("Total Views")
plt.title("Top 10 Artists by Total Views")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

distribution of song popularity using a BOX PLOT

```python
plt.figure(figsize=(8, 5))
plt.boxplot(df["Views"].dropna())

plt.ylabel("Number of Views")
plt.title("Distribution of Song Views")

plt.show()
```
